```python
# =========================
# 1. KÜTÜPHANELER
# =========================
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Dropout
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

from sklearn.metrics import confusion_matrix, classification_report

print("TensorFlow version:", tf.__version__)

# =========================
# 2. MNIST VERİ SETİ
# =========================
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test  = x_test.astype("float32") / 255.0

# Shape: (N, 28, 28, 1)
x_train = np.expand_dims(x_train, -1)
x_test  = np.expand_dims(x_test, -1)

# One-hot
y_train_cat = to_categorical(y_train, 10)
y_test_cat  = to_categorical(y_test, 10)

# =========================
# 3. STM32 UYUMLU CNN (BN YOK)
# =========================
model = Sequential([
    Conv2D(8, (3,3), activation='relu', input_shape=(28,28,1)),
    MaxPooling2D(2,2),

    Conv2D(16, (3,3), activation='relu'),
    MaxPooling2D(2,2),

    Flatten(),
    Dense(32, activation='relu'),
    Dense(10, activation='softmax')
])

model.summary()

# =========================
# 4. DERLEME
# =========================
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# =========================
# 5. EĞİTİM
# =========================
callbacks = [
    EarlyStopping(patience=3, restore_best_weights=True),
    ModelCheckpoint("mnist_cnn.h5", save_best_only=True)
]

history = model.fit(
    x_train, y_train_cat,
    epochs=20,
    batch_size=64,
    validation_split=0.1,
    callbacks=callbacks,
    verbose=1
)

# =========================
# 6. TEST SONUÇLARI
# =========================
test_loss, test_acc = model.evaluate(x_test, y_test_cat, verbose=0)
print(f"\nTest Accuracy: {test_acc*100:.2f}%")

# =========================
# 7. CONFUSION MATRIX
# =========================
y_pred = model.predict(x_test)
y_pred_classes = np.argmax(y_pred, axis=1)

cm = confusion_matrix(y_test, y_pred_classes)

plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix - MNIST")
plt.show()

# =========================
# 8. CLASSIFICATION REPORT
# =========================
print("\nClassification Report:")
print(classification_report(y_test, y_pred_classes))

# =========================
# 9. TFLITE DÖNÜŞÜM (STM32 İÇİN)
# =========================
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

with open("mnist_cnn.tflite", "wb") as f:
    f.write(tflite_model)

print("\nTFLite model created: mnist_cnn.tflite")
```

**SqueezeNet**

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, GlobalAveragePooling2D, Dense, Concatenate
from tensorflow.keras.models import Model

print("--- SqueezeNet İşlemleri Başlıyor (Düzeltilmiş Versiyon) ---")

# =========================================================
# 1. VERİ HAZIRLIĞI
# =========================================================
print("Veri yükleniyor...")
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()

# Etiketleri burada One-Hot yapıyoruz (Map fonksiyonunda hata çıkmaması için)
train_labels = tf.keras.utils.to_categorical(train_labels, 10)
test_labels = tf.keras.utils.to_categorical(test_labels, 10)

# Veri Ön İşleme Fonksiyonu (Tamamen TensorFlow fonksiyonları ile)
def preprocess_data(images, labels):
    # np.expand_dims yerine tf.expand_dims kullanıyoruz:
    images = tf.expand_dims(images, axis=-1)
    images = tf.image.convert_image_dtype(images, tf.float32)
    images = tf.image.resize(images, (32, 32)) # 32x32 boyut
    images = tf.image.grayscale_to_rgb(images) # 3 Kanal (RGB)
    return images, labels

# Pipeline Oluşturma
train_ds = tf.data.Dataset.from_tensor_slices((train_images, train_labels))
train_ds = train_ds.map(preprocess_data).batch(32)

# =========================================================
# 2. MODEL MİMARİSİ (SqueezeNet Mini)
# =========================================================
def fire_module(x, s1x1, e1x1, e3x3):
    squeeze = Conv2D(s1x1, (1, 1), activation='relu')(x)
    expand1x1 = Conv2D(e1x1, (1, 1), activation='relu')(squeeze)
    expand3x3 = Conv2D(e3x3, (3, 3), padding='same', activation='relu')(squeeze)
    return Concatenate()([expand1x1, expand3x3])

def build_squeezenet():
    input_shape = (32, 32, 3)
    inputs = Input(shape=input_shape)

    x = Conv2D(32, (3, 3), strides=(2, 2), padding='same', activation='relu')(inputs)
    x = MaxPooling2D(pool_size=(2, 2))(x)

    x = fire_module(x, 8, 32, 32)
    x = fire_module(x, 8, 32, 32)
    x = MaxPooling2D(pool_size=(2, 2))(x)
    x = fire_module(x, 16, 64, 64)

    x = GlobalAveragePooling2D()(x)
    outputs = Dense(10, activation='softmax')(x)

    return Model(inputs, outputs)

model = build_squeezenet()
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# =========================================================
# 3. EĞİTİM
# =========================================================
print("SqueezeNet eğitiliyor...")
model.fit(train_ds, epochs=5)

# =========================================================
# 4. TFLITE DÖNÜŞÜM VE KAYIT
# =========================================================
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

with open("mnist_squeezenet.tflite", "wb") as f:
    f.write(tflite_model)
print("\nBAŞARILI: 'mnist_squeezenet.tflite' oluşturuldu. Soldaki menüden indirebilirsin.")

# =========================================================
# 5. C KODU İÇİN TEST GÖRÜNTÜSÜ (32x32 RGB)
# =========================================================
print("\n--- AŞAĞIDAKİ KODU KOPYALA VE MAIN.C İÇİNE YAPIŞTIR ---")
# Rastgele bir '7' rakamı buluyoruz (Orijinal test setinden)
(_, _), (x_test_raw, y_test_raw_labels) = tf.keras.datasets.mnist.load_data()
idx = np.where(y_test_raw_labels == 7)[0][0]
img = x_test_raw[idx]

# İşlem (Numpy kullanabiliriz çünkü dataset içinde değiliz)
img = img.astype("float32") / 255.0
img = np.expand_dims(img, axis=-1)
img = tf.image.resize(img, (32, 32))
img = tf.image.grayscale_to_rgb(img)

print(f"/* SqueezeNet Test Image (Label: 7) */")
print(f"const float test_image_7_sq[32*32*3] = {{")
flat_img = img.numpy().flatten()
for i, val in enumerate(flat_img):
    print(f"{val:.4f}f, ", end="")
    if (i+1) % 10 == 0: print()
print("\n};")
```

**EfficientNet**

```python
# =========================================================
#  2. EFFICIENTNET EĞİTİM, CONFUSION MATRIX VE ÇIKTI KODU
# =========================================================
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.layers import Input, GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model

print("--- EfficientNetB0 İşlemleri Başlıyor ---")

# --- A. Veri Hazırlığı (32x32 RGB) ---
print("Veri yükleniyor...")
(train_images, train_labels), (test_images, test_labels_raw) = tf.keras.datasets.mnist.load_data()

# One-Hot Encoding (Eğitim için)
train_labels_cat = tf.keras.utils.to_categorical(train_labels, 10)
test_labels_cat = tf.keras.utils.to_categorical(test_labels_raw, 10)

# Ön İşleme Fonksiyonu
def preprocess_data(images, labels):
    images = tf.expand_dims(images, axis=-1)
    images = tf.image.convert_image_dtype(images, tf.float32)
    images = tf.image.resize(images, (32, 32)) # EfficientNet için min 32x32
    images = tf.image.grayscale_to_rgb(images) # RGB şart
    return images, labels

# Pipeline
train_ds = tf.data.Dataset.from_tensor_slices((train_images, train_labels_cat))
train_ds = train_ds.map(preprocess_data).batch(32)

test_ds = tf.data.Dataset.from_tensor_slices((test_images, test_labels_cat))
test_ds = test_ds.map(preprocess_data).batch(32)

# --- B. EfficientNetB0 Modeli ---
def build_efficientnet():
    input_shape = (32, 32, 3)
    inputs = Input(shape=input_shape)

    # weights=None diyerek sıfırdan eğitiyoruz (ImageNet ağırlıkları 32x32'de sorun çıkarabilir)
    base_model = tf.keras.applications.EfficientNetB0(
        input_shape=input_shape,
        include_top=False,
        weights=None
    )

    x = base_model(inputs)
    x = GlobalAveragePooling2D()(x)
    outputs = Dense(10, activation='softmax')(x)

    return Model(inputs, outputs)

model = build_efficientnet()
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# --- C. Eğitim ---
print("EfficientNetB0 eğitiliyor (Bu biraz uzun sürebilir)...")
history = model.fit(train_ds, epochs=3, validation_data=test_ds) # Hız için 3 epoch, artırabilirsin

# --- D. Confusion Matrix ve Rapor ---
print("\n--- Test Sonuçları ve Confusion Matrix ---")

# Tahminleri al
y_pred_probs = model.predict(test_ds)
y_pred_classes = np.argmax(y_pred_probs, axis=1)

# Gerçek etiketler (Dataset sırasını koruduğu için test_labels_raw kullanabiliriz)
# Ancak batch sırasında kayma olmaması için en güvenlisi test_labels_raw'u kullanmaktır.
cm = confusion_matrix(test_labels_raw, y_pred_classes)

# Çizdir
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=range(10), yticklabels=range(10))
plt.xlabel('Tahmin Edilen (Predicted)')
plt.ylabel('Gerçek (True)')
plt.title('Confusion Matrix - EfficientNetB0')
plt.show()

# Sınıflandırma Raporu
print("\nClassification Report:")
print(classification_report(test_labels_raw, y_pred_classes))

# --- E. TFLite Dönüştürme ---
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

with open("mnist_efficientnet.tflite", "wb") as f:
    f.write(tflite_model)
print("\nBAŞARILI: 'mnist_efficientnet.tflite' oluşturuldu.")

# --- F. STM32 İçin C Test Dizisi ---
# (Eğer model sığarsa kullanmak için)
print("\n--- C KODU İÇİN TEST GÖRÜNTÜSÜ ---")
idx = np.where(test_labels_raw == 7)[0][0]
img = test_images[idx]

img = img.astype("float32") / 255.0
img = np.expand_dims(img, axis=-1)
img = tf.image.resize(img, (32, 32))
img = tf.image.grayscale_to_rgb(img)

print(f"/* EfficientNet Test Image (Label: 7) */")
print(f"const float test_image_7_eff[32*32*3] = {{")
flat_img = img.numpy().flatten()
for i, val in enumerate(flat_img):
    print(f"{val:.4f}f, ", end="")
    if (i+1) % 10 == 0: print()
print("\n};")
```

**MobileNetV2**

```python
# =========================================================
#  3. MOBILENETV2 (ALPHA=0.35) EĞİTİM VE ÇIKTI KODU
# =========================================================
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.layers import Input, GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model

print("--- MobileNetV2 (Mini) İşlemleri Başlıyor ---")

# --- A. Veri Hazırlığı (32x32 RGB) ---
print("Veri yükleniyor...")
(train_images, train_labels), (test_images, test_labels_raw) = tf.keras.datasets.mnist.load_data()

# One-Hot Encoding
train_labels_cat = tf.keras.utils.to_categorical(train_labels, 10)
test_labels_cat = tf.keras.utils.to_categorical(test_labels_raw, 10)

def preprocess_data(images, labels):
    images = tf.expand_dims(images, axis=-1)
    images = tf.image.convert_image_dtype(images, tf.float32)
    images = tf.image.resize(images, (32, 32)) # MobileNet min 32x32 ister
    images = tf.image.grayscale_to_rgb(images) # RGB olmak zorunda
    return images, labels

# Pipeline
train_ds = tf.data.Dataset.from_tensor_slices((train_images, train_labels_cat))
train_ds = train_ds.map(preprocess_data).batch(32)

test_ds = tf.data.Dataset.from_tensor_slices((test_images, test_labels_cat))
test_ds = test_ds.map(preprocess_data).batch(32)

# --- B. MobileNetV2 Modeli (Optimize Edilmiş) ---
def build_mobilenet():
    input_shape = (32, 32, 3)
    inputs = Input(shape=input_shape)

    # KRİTİK NOKTA: alpha=0.35 diyerek modeli %65 küçültüyoruz.
    # Bu sayede STM32 Flash hafızasına sığıyor.
    base_model = tf.keras.applications.MobileNetV2(
        input_shape=input_shape,
        include_top=False,
        weights=None,
        alpha=0.35
    )

    x = base_model(inputs)
    x = GlobalAveragePooling2D()(x)
    outputs = Dense(10, activation='softmax')(x)

    return Model(inputs, outputs)

model = build_mobilenet()
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# --- C. Eğitim ---
print("MobileNetV2 eğitiliyor...")
history = model.fit(train_ds, epochs=4, validation_data=test_ds)

# --- D. Confusion Matrix ve Rapor ---
print("\n--- Test Sonuçları ve Confusion Matrix ---")
y_pred_probs = model.predict(test_ds)
y_pred_classes = np.argmax(y_pred_probs, axis=1)

cm = confusion_matrix(test_labels_raw, y_pred_classes)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', xticklabels=range(10), yticklabels=range(10))
plt.xlabel('Tahmin Edilen (Predicted)')
plt.ylabel('Gerçek (True)')
plt.title('Confusion Matrix - MobileNetV2 (alpha=0.35)')
plt.show()

print("\nClassification Report:")
print(classification_report(test_labels_raw, y_pred_classes))

# --- E. TFLite Dönüştürme ---
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

with open("mnist_mobilenet.tflite", "wb") as f:
    f.write(tflite_model)
print("\nBAŞARILI: 'mnist_mobilenet.tflite' oluşturuldu.")

# --- F. STM32 İçin C Test Dizisi ---
print("\n--- C KODU İÇİN TEST GÖRÜNTÜSÜ ---")
idx = np.where(test_labels_raw == 7)[0][0] # Yine 7 rakamını alalım
img = test_images[idx]

img = img.astype("float32") / 255.0
img = np.expand_dims(img, axis=-1)
img = tf.image.resize(img, (32, 32))
img = tf.image.grayscale_to_rgb(img)

print(f"/* MobileNet Test Image (Label: 7) */")
print(f"const float test_image_7_mob[32*32*3] = {{")
flat_img = img.numpy().flatten()
for i, val in enumerate(flat_img):
    print(f"{val:.4f}f, ", end="")
    if (i+1) % 10 == 0: print()
print("\n};")
```

**ShuffleNet**

```python
# =========================================================
#  4. SHUFFLENET (V2-Like) EĞİTİM VE ÇIKTI KODU
# =========================================================
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.layers import Input, Conv2D, DepthwiseConv2D, Dense, GlobalAveragePooling2D, BatchNormalization, Activation, Concatenate, Reshape, Permute
from tensorflow.keras.models import Model

print("--- ShuffleNet İşlemleri Başlıyor ---")

# --- A. Veri Hazırlığı (32x32 RGB) ---
print("Veri yükleniyor...")
(train_images, train_labels), (test_images, test_labels_raw) = tf.keras.datasets.mnist.load_data()

# One-Hot Encoding
train_labels_cat = tf.keras.utils.to_categorical(train_labels, 10)
test_labels_cat = tf.keras.utils.to_categorical(test_labels_raw, 10)

def preprocess_data(images, labels):
    images = tf.expand_dims(images, axis=-1)
    images = tf.image.convert_image_dtype(images, tf.float32)
    images = tf.image.resize(images, (32, 32)) # ShuffleNet min 32x32
    images = tf.image.grayscale_to_rgb(images) # RGB
    return images, labels

# Pipeline
train_ds = tf.data.Dataset.from_tensor_slices((train_images, train_labels_cat))
train_ds = train_ds.map(preprocess_data).batch(32)

test_ds = tf.data.Dataset.from_tensor_slices((test_images, test_labels_cat))
test_ds = test_ds.map(preprocess_data).batch(32)

# --- B. ShuffleNet Mimarisi (Basitleştirilmiş) ---
def channel_shuffle(x, groups):
    # Kanalları karıştırma işlemi
    height, width, channels = x.shape[1], x.shape[2], x.shape[3]
    channels_per_group = channels // groups

    x = Reshape((height, width, groups, channels_per_group))(x)
    x = Permute((1, 2, 4, 3))(x) # Grupları karıştır
    x = Reshape((height, width, channels))(x)
    return x

def shuffle_unit(x, groups, channels):
    # 1x1 Group Conv (Giriş)
    x = Conv2D(channels, (1, 1), padding='same', use_bias=False)(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)

    # Channel Shuffle
    x = channel_shuffle(x, groups)

    # 3x3 Depthwise Conv (Maliyet düşürücü)
    x = DepthwiseConv2D((3, 3), padding='same', use_bias=False)(x)
    x = BatchNormalization()(x)

    # 1x1 Group Conv (Çıkış)
    x = Conv2D(channels, (1, 1), padding='same', use_bias=False)(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    return x

def build_shufflenet():
    input_shape = (32, 32, 3)
    inputs = Input(shape=input_shape)

    # Giriş Katmanı
    x = Conv2D(24, (3, 3), strides=(2, 2), padding='same', activation='relu')(inputs)
    x = tf.keras.layers.MaxPooling2D(pool_size=(2, 2))(x)

    # Shuffle Üniteleri (Tekrarlı)
    x = shuffle_unit(x, groups=2, channels=48)
    x = shuffle_unit(x, groups=2, channels=48)
    x = tf.keras.layers.MaxPooling2D(pool_size=(2, 2))(x)
    x = shuffle_unit(x, groups=2, channels=96)

    # Çıkış
    x = GlobalAveragePooling2D()(x)
    outputs = Dense(10, activation='softmax')(x)

    return Model(inputs, outputs)

model = build_shufflenet()
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# --- C. Eğitim ---
print("ShuffleNet eğitiliyor...")
history = model.fit(train_ds, epochs=5, validation_data=test_ds)

# --- D. Confusion Matrix ve Rapor ---
print("\n--- Test Sonuçları ve Confusion Matrix ---")
y_pred_probs = model.predict(test_ds)
y_pred_classes = np.argmax(y_pred_probs, axis=1)

cm = confusion_matrix(test_labels_raw, y_pred_classes)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges', xticklabels=range(10), yticklabels=range(10))
plt.xlabel('Tahmin Edilen (Predicted)')
plt.ylabel('Gerçek (True)')
plt.title('Confusion Matrix - ShuffleNet')
plt.show()

print("\nClassification Report:")
print(classification_report(test_labels_raw, y_pred_classes))

# --- E. TFLite Dönüştürme ---
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

with open("mnist_shufflenet.tflite", "wb") as f:
    f.write(tflite_model)
print("\nBAŞARILI: 'mnist_shufflenet.tflite' oluşturuldu.")

# --- F. STM32 İçin C Test Dizisi ---
print("\n--- C KODU İÇİN TEST GÖRÜNTÜSÜ ---")
idx = np.where(test_labels_raw == 7)[0][0]
img = test_images[idx]

img = img.astype("float32") / 255.0
img = np.expand_dims(img, axis=-1)
img = tf.image.resize(img, (32, 32))
img = tf.image.grayscale_to_rgb(img)

print(f"/* ShuffleNet Test Image (Label: 7) */")
print(f"const float test_image_7_shuff[32*32*3] = {{")
flat_img = img.numpy().flatten()
for i, val in enumerate(flat_img):
    print(f"{val:.4f}f, ", end="")
    if (i+1) % 10 == 0: print()
print("\n};")
```
